## Task 2 — Calculation
**Objective:** Calculate the percentage of elevators with an expired license (LICENSE EXPIRY DATE before today).

**Relevant data facts:**
- Total records: 45,383
- Today's date: May 15, 2026
- All 45,383 records have a LICENSEEXPIRYDATE before May 15, 2026
- Records with LICENSE STATUS of EXPIRED: 68 (0.15%)

### Technique 1 — Zero-Shot

**Prompt:**

> The Ontario elevator registry contains 45,383 records, each with a LICENSEEXPIRYDATE. Today's date is May 15, 2026. All 45,383 records have a LICENSEEXPIRYDATE before today.
>
> Calculate the percentage of elevators with an expired license (LICENSEEXPIRYDATE before today).

---

**Output:**

To calculate the percentage of elevators with an expired license:

**Given:**
- Total records: 45,383
- Records with LICENSEEXPIRYDATE before May 15, 2026: 45,383

**Calculation:**
45,383 ÷ 45,383 × 100 = **100%**

**100% of elevators in the dataset have a license expiry date before today.**

This is an unusual result worth noting. A 100% expiry rate does not necessarily mean the entire fleet is non-operational. The dataset appears to have been collected around 2017, meaning all recorded expiry dates now fall before May 2026 regardless of the elevator's actual current license status. The LICENSEEXPIRYDATE column reflects when the license was last recorded — not whether it has since been renewed. For a more operationally meaningful measure, filtering by LICENSE STATUS = EXPIRED returns only 68 records (0.15% of the dataset).

### Technique 2 — Few-Shot

**Prompt:**

> Here are some examples of how to calculate an expiry percentage:
> - 10 records expire before today out of 50 total → 10 ÷ 50 × 100 = 20% expired
> - 3 records expire before today out of 10 total → 3 ÷ 10 × 100 = 30% expired
> - 0 records expire before today out of 25 total → 0 ÷ 25 × 100 = 0% expired
>
> Using the same method: 45,383 records out of 45,383 total have a LICENSEEXPIRYDATE before May 15, 2026. Calculate the percentage of expired licenses and interpret the result.

---

**Output:**

Following the same method:

45,383 ÷ 45,383 × 100 = **100%**

**100% of elevator records have a license expiry date before today.**

This follows the pattern from the examples but represents an extreme edge case. Unlike the examples where a subset of records expired, here every single record meets the condition. This does not necessarily reflect the operational reality of the fleet — it is a consequence of dataset age. The data was recorded around 2017, so all expiry dates now predate May 2026 by several years regardless of whether a license was subsequently renewed.

A more meaningful distinction separates two ways of defining "expired":
- **By date (LICENSEEXPIRYDATE before today):** 45,383 records — 100%
- **By status (LICENSE STATUS = EXPIRED):** 68 records — 0.15%

The date-based figure is technically correct but operationally misleading for this dataset. The status-based figure is the more reliable indicator of which elevators actually have lapsed licenses.

### Technique 3 — Chain-of-Thought

**Prompt:**

> The Ontario elevator registry contains 45,383 records, each with a LICENSEEXPIRYDATE column. Today's date is May 15, 2026.
>
> Work through this step by step: first define what "expired" means in this context, then determine how many records meet that condition, then calculate the percentage, then interpret what that result means for the dataset and whether it is a useful metric.

---

**Output:**

**Step 1 — Define "expired"**
In this context, an expired license is one where the LICENSEEXPIRYDATE falls before today, May 15, 2026. This is a purely date-based definition — it does not consider whether the license was subsequently renewed or what the current LICENSE STATUS is.

**Step 2 — Determine how many records meet the condition**
All 45,383 records have a LICENSEEXPIRYDATE before May 15, 2026. The most recent expiry dates in the dataset fall around 2017–2018, meaning every record in the dataset is past its recorded expiry date relative to today.

**Step 3 — Calculate the percentage**
45,383 ÷ 45,383 × 100 = **100%**

**Step 4 — Interpret the result and assess usefulness**
A 100% expiry rate sounds alarming but is a product of dataset age, not a genuine operational crisis. The data was collected circa 2017 — all expiry dates now precede 2026 by several years. This makes LICENSEEXPIRYDATE an unreliable standalone metric for determining whether a license is currently valid.

This metric is **not useful as defined** for this dataset. A more meaningful approach distinguishes two definitions of "expired":

- **By date (LICENSEEXPIRYDATE before today):** 45,383 records — 100% — reflects dataset age, not operational status
- **By status (LICENSE STATUS = EXPIRED):** 68 records — 0.15% — reflects elevators the registry explicitly considers to have lapsed licenses

For any operational analysis, the status-based figure should be used. The date-based percentage only becomes meaningful if the dataset is kept current with up-to-date expiry records.

## Comparison — Which Technique Worked Best and Why?

---

### Task 1 — Classification

**Best technique: Chain-of-Thought**

Chain-of-thought produced the most defensible classifications because it required explicit reasoning before each decision. For straightforward statuses like ACTIVE and TERMINATED, all three techniques agreed. The difference appeared on ambiguous values — specifically `ENTERED`, `BY REQUEST`, and `HOLD_TSD`. Chain-of-thought was the only technique that flagged `ENTERED` as *low confidence*, noting it was a transitional status with an ambiguous service state and only 4 occurrences in the dataset. Zero-shot and few-shot both classified it as Operational without any caveat. For `HOLD_TSD`, chain-of-thought explicitly named the Technical Standards Division and described the regulatory mechanism — giving the classification a traceable justification rather than a pattern match.

**Where a technique did not add value: Few-Shot**

The three examples provided (ACTIVE, TERMINATED, EXPIRED) were the most obvious cases in the dataset — values that any technique would classify correctly without guidance. The examples gave the model a pattern to follow but did not help with the genuinely hard cases like `HOLD_TSD` or `BY REQUEST`, since nothing in the examples resembled a regulatory hold or conditional suspension. Few-shot constrained the model to mimicking the example format rather than reasoning through harder cases independently.

---

### Task 2 — Calculation

**Best technique: Chain-of-Thought**

All three techniques arrived at the same numerical answer (100%), so accuracy was not the differentiator. What separated them was depth of interpretation. Zero-shot noted the dataset age issue briefly. Few-shot noted it as an "extreme edge case." Chain-of-thought went furthest — it explicitly questioned whether the metric was useful at all, named the specific problem (LICENSEEXPIRYDATE reflects when data was recorded, not current status), and drew a clear distinction between two definitions of "expired": 100% by date vs. 0.15% by status. Only chain-of-thought recommended which figure should actually be used for operational analysis.

**Where a technique did not add value: Few-Shot**

Task 2 is a single arithmetic operation. Providing three examples of percentage calculations added no accuracy benefit — the model can divide two numbers without demonstration. The examples made the prompt longer and the output marginally longer without producing any insight that zero-shot did not also produce. This is the clearest case in the lab where few-shot added no value.

---

### Task 3 — Open-Ended Analysis

**Best technique: Chain-of-Thought**

By structuring the analysis into four explicit steps — concentration, clusters, economic factors, takeaways — chain-of-thought produced the most complete and layered response. It was the only technique to:
- Identify the 1998 Toronto amalgamation as the reason Scarborough, Etobicoke, and North York appear as separate cities, and note that GTA counts are understated under "Toronto" alone
- Frame northern Ontario's absence as a meaningful finding in itself rather than just an observation
- Produce a numbered takeaway list that directly addressed what the data means for analysis — specifically that the GTA cluster skews any province-wide metric

Zero-shot produced solid observations but stayed closer to the surface of the numbers. Few-shot shaped the style of observations to match the examples but did not go deeper.

**Where a technique did not add value: Few-Shot**

For an open-ended analysis task, the examples defined the *style* of the answer but constrained its depth. The few-shot output followed the example format closely — bold lead-in followed by supporting data — but did not venture beyond what zero-shot also identified. On open-ended tasks, showing the model what kind of observation to make can actually narrow the response by anchoring it to the example pattern rather than letting the model explore the data freely.

---

### Overall Summary

| Task | Best Technique | Technique That Added Least Value |
|---|---|---|
| Classification | Chain-of-Thought | Few-Shot |
| Calculation | Chain-of-Thought | Few-Shot |
| Open-Ended Analysis | Chain-of-Thought | Few-Shot |

Chain-of-thought was the strongest technique across all three tasks. The common reason: it forced the model to make its reasoning visible before committing to an answer. This produced traceable justifications for ambiguous classifications (Task 1), critical evaluation of a misleading metric (Task 2), and layered analysis that went beyond reading the numbers (Task 3).

Few-shot consistently added the least value in this lab. It is most effective when the model needs to learn an unfamiliar output format — but all three tasks had clear expected formats that the model produced correctly without examples. When the format is already obvious, examples become overhead rather than guidance. The clearest evidence: few-shot on Task 2 produced no insight that zero-shot did not also produce, despite a longer prompt.
