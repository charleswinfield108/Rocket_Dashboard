## Task 2 — Calculation
**Objective:** Calculate the percentage of elevators with an expired license (LICENSE EXPIRY DATE before today).

**Relevant data facts:**
- Total records: 45,383
- Today's date: May 15, 2026
- All 45,383 records have a LICENSEEXPIRYDATE before May 15, 2026
- Records with LICENSE STATUS of EXPIRED: 68 (0.15%)

### Technique 1 — Zero-Shot

**Prompt:**

> The Ontario elevator registry contains 45,383 records, each with a LICENSEEXPIRYDATE. Today's date is May 15, 2026. All 45,383 records have a LICENSEEXPIRYDATE before today.
>
> Calculate the percentage of elevators with an expired license (LICENSEEXPIRYDATE before today).

---

**Output:**

To calculate the percentage of elevators with an expired license:

- Records with LICENSEEXPIRYDATE before May 15, 2026: **45,383**
- Total records: **45,383**
- Percentage: 45,383 ÷ 45,383 × 100 = **100%**

**100% of elevators in the dataset have a license expiry date before today.**

This result is notable — it likely reflects the age of the dataset rather than a true operational picture. The data appears to have been collected around 2017, meaning all expiry dates now fall in the past regardless of the elevator's current license status.

### Technique 2 — Few-Shot

**Prompt:**

> Here are some examples of how to calculate an expiry percentage:
> - 10 records expire before today out of 50 total → 10 ÷ 50 × 100 = 20% expired
> - 3 records expire before today out of 10 total → 3 ÷ 10 × 100 = 30% expired
> - 0 records expire before today out of 25 total → 0 ÷ 25 × 100 = 0% expired
>
> Using the same method: 45,383 records out of 45,383 total have a LICENSEEXPIRYDATE before May 15, 2026. Calculate the percentage of expired licenses and interpret the result.

---

**Output:**

Following the same method:

45,383 ÷ 45,383 × 100 = **100%**

**100% of elevator records have a license expiry date before today.**

This matches the pattern from the examples but represents an extreme case — a 100% expiry rate. This does not necessarily mean every elevator is non-operational. The LICENSEEXPIRYDATE reflects when the license was last recorded in the dataset, and since the data is from approximately 2017, all dates now predate May 2026. A more operationally meaningful measure would be to filter by LICENSE STATUS = EXPIRED, which accounts for only 68 records (0.15%).

### Technique 3 — Chain-of-Thought

**Prompt:**

> The Ontario elevator registry contains 45,383 records, each with a LICENSEEXPIRYDATE column. Today's date is May 15, 2026.
>
> Work through this step by step: first define what "expired" means in this context, then determine how many records meet that condition, then calculate the percentage, then interpret what that result means for the dataset and whether it is a useful metric.

---

**Output:**

**Step 1 — Define "expired"**  
A license is expired if its LICENSEEXPIRYDATE falls before today, May 15, 2026. This is a date-based definition and does not consider the LICENSE STATUS field.

**Step 2 — Count qualifying records**  
All 45,383 records have a LICENSEEXPIRYDATE before May 15, 2026. The most recent dates in the dataset fall around 2017–2018, meaning every record is technically past its expiry date relative to today.

**Step 3 — Calculate the percentage**  
45,383 ÷ 45,383 × 100 = **100%**

**Step 4 — Interpret the result**  
A 100% expiry rate is a product of dataset age, not operational reality. The data was collected circa 2017, so all expiry dates now precede 2026 by several years. Using LICENSEEXPIRYDATE as the sole measure of expiry is misleading for this dataset.

A more meaningful interpretation distinguishes two definitions of "expired":
- **By date:** 45,383 records (100%) — every elevator's recorded expiry date is in the past
- **By status:** 68 records (0.15%) — elevators explicitly marked EXPIRED in the LICENSE STATUS field

For operational analysis, the status-based figure is more reliable.

## Comparison — Which Technique Worked Best and Why?

---

### Task 1 — Classification

**Best technique: Chain-of-Thought**

Chain-of-thought produced the most defensible classifications because it forced explicit reasoning before each decision. For straightforward statuses like ACTIVE and TERMINATED, all three techniques agreed. But for ambiguous values — specifically `HOLD_TSD`, `BY REQUEST`, and `ENTERED` — the step-by-step process made the reasoning traceable and justified. Chain-of-thought was the only technique that flagged `ENTERED` as *low confidence*, acknowledging that a transitional status resists a clean binary classification. Zero-shot and few-shot both classified it as Operational without any caveat.

**Where a technique did not add value: Few-Shot**

The three examples provided (ACTIVE, TERMINATED, EXPIRED) were the most obvious cases in the dataset — values that any technique would classify correctly without guidance. The examples did not help with the genuinely ambiguous values like `HOLD_TSD` or `BY REQUEST` because nothing in the examples resembled a regulatory hold or conditional suspension. Few-shot constrained the model to pattern-matching against simple examples rather than reasoning through harder cases.

---

### Task 2 — Calculation

**Best technique: Chain-of-Thought**

All three techniques arrived at the same numerical answer (100%), so accuracy was not the differentiator. What chain-of-thought added was critical interpretation. By forcing a step on "is this a useful metric?", it was the only technique that explicitly identified the core problem with the result — that 100% expiry is an artifact of dataset age, not a meaningful operational finding. It also distinguished between two definitions of "expired" (by date vs. by status) and recommended the status-based figure (0.15%) as more reliable. Zero-shot and few-shot reported the result without questioning its validity.

**Where a technique did not add value: Few-Shot**

Task 2 is a single arithmetic operation — divide two numbers and multiply by 100. Providing examples of simpler calculations (10/50, 3/10) added no accuracy benefit since the model can perform basic division without demonstration. The examples made the prompt longer without improving the output. This is the clearest case in the lab where few-shot added no value.

---

### Task 3 — Open-Ended Analysis

**Best technique: Chain-of-Thought**

By structuring the analysis into four explicit steps — concentration, clusters, economic factors, takeaways — chain-of-thought produced insights that neither of the other techniques reached. Notably, it identified that Scarborough, Etobicoke, and North York appear as separate cities in the data due to Toronto's 1998 amalgamation, which explains an otherwise puzzling feature of the dataset. It also explicitly named the absence of northern Ontario cities as an infrastructure gap — an observation that requires reasoning beyond the numbers, not just reading the table.

**Where a technique did not add value: Few-Shot**

For an open-ended analysis task, few-shot examples define the *style* of the answer rather than improving its depth or accuracy. The output followed the example format closely but did not go further than zero-shot in terms of insight. On open-ended tasks, showing the model what kind of observation to make risks narrowing its response to mimic the examples rather than exploring the data freely. Zero-shot actually surfaced the GTA cluster point without being constrained by the example format.

---

### Overall Summary

| Task | Best Technique | Technique That Added Least Value |
|---|---|---|
| Classification | Chain-of-Thought | Few-Shot |
| Calculation | Chain-of-Thought | Few-Shot |
| Open-Ended Analysis | Chain-of-Thought | Few-Shot |

Chain-of-thought was the strongest technique across all three tasks because it forced the model to make its reasoning visible and to question its own outputs — particularly valuable when results are ambiguous (Task 1) or misleading on the surface (Task 2).

Few-shot consistently added the least value in this lab. It is most useful when the model needs to learn an unfamiliar output format or style — but all three tasks here had straightforward expected formats. When the output format is already clear, examples become overhead rather than guidance.
